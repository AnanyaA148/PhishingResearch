# Spam Ham Data

```python
import pandas as pd


df = pd.read_csv('Phishing_Emails/spamhamdata.csv', usecols=[0, 1]) #Spamhamdata
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

sns.countplot(data=df, x="label")
plt.show()
```

```python
textSpam = ""
textHam = ""
for row in df.itertuples():
    #print(row.label)
    if (row.label == "spam"):
        #print(row.body)
        textSpam = textSpam + row.body
    else:
        textHam = textHam + row.body

```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wc = WordCloud(
    width=1200,
    height=600,
    background_color="white",
    max_words=200,
    colormap="viridis"
)

wordcloud = wc.generate(textSpam)

plt.figure(figsize=(15, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```
